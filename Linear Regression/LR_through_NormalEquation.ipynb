implementing simple linear regression through normal equation

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.loadtxt("ex2x.dat", dtype=np.float32)
y = np.loadtxt("ex2y.dat", dtype=np.float32)

print(x.shape)
print(y.shape)

print(x)
print(y)
```

```python
plt.scatter(x, y, marker ="+", c = "r")
plt.xlabel("x values")
plt.ylabel("y values")
plt.title("x against y")
```

```python
# add a coulumn of ones to the x matrix

x = np.column_stack((np.ones(len(x)), x))
```

```python
#implement the equation θ=((XT)X)^−1*(XT)y

theta = np.linalg.pinv(x.T@ x) @ x.T @ y
theta
```

```python
plt.scatter(x[:,1], y, marker ="+", c = "r")
plt.plot(x, theta[0] + theta[1] * x)
plt.xlabel("x values")
plt.ylabel("y values")
plt.title("x against y")
```
