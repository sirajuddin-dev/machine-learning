```python
# import libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
#load data
x = np.loadtxt('ex2x.dat', dtype = np.float32)
y = np.loadtxt('ex2y.dat', dtype = np.float32)
```

```python
print(x)
print(y)
```

```python
#initialize w and a
w = 1
a = 1
m = len(x)
```

```python
#the line equation
def hypothesis(x, w, a):
    return w*x + a
```

```python
#plot  or line against the data with intial values of w and a
plt.scatter(x, y, marker="+", c = "r")
plt.plot(x, hypothesis(x, w, a))
```

```python
#fn to calculate cost
def cost(x, w, a, y):
    return (sum(hypothesis(x, w, a) - y)**2) / (2*m)
```

```python
#define separate gradients for w and a to update indisually 
def gradient_descent_w(x, w, a, y):
    return sum((hypothesis(x, w, a) - y) * x) / m

def gradient_descent_a(x, w, a, y):
    return sum(hypothesis(x, w, a) - y) / m
```

```python
#calculate the iteration and cost value to plot later
costVal = []
epochs = []
for i in range(1700):
    costVal.append(cost(x, w, a, y))
    #update values
    alpha = 0.02
    tempw = w - alpha * gradient_descent_w(x, w, a, y)
    tempa = a - alpha * gradient_descent_a(x, w, a, y)
    w = tempw
    a = tempa
    epochs.append(i)
    
```

```python
# line against dat when gradient decent ran
plt.scatter(x, y, marker="+", c = "r")
plt.plot(x, hypothesis(x, w, a))
```

```python
#plot cost against epochs
plt.plot(epochs, costVal)
plt.xlabel("iteration")
plt.ylabel("cost value")
plt.title("cost against iteration")
```
